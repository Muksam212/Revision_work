# VIF and Multicollinearity

- VIF: Variance Inflation Factor measure how much the variance of an estimated regression coefficient is increased because of correlation among predictor variables in a model

- Multicollinearity: This happens when independent variables in a regression model are linearly correlated with each other. It does not reduce overall predictive power, but it makes coefficient estimates unstable and standard errors large.


How does VIF work?
- It calculates an R² value by regressing a single predictor variable against all other predictor variables in the model. The formula is:

$$
VIF = \frac{1}{1 - R^2}
$$


Interpreting the VIF Rules

- VIF = 1: The predictor does not correlate with other variables.
- VIF between 1 and 5: Low or moderate correlation, which is generally acceptable.
- VIF between 5 and 10: Moderate to high correlation that may warrant caution.
- VIF > 10: High multicollinearity that severely inflates standard errors and often requires action.

# How to fix multicollinearity

- To drop the columns
- Combine variables
- Add data
- Use alternative method: PCA (Principal Component Analysis) or Regularized Regression (Lasso, Ridge)

# Implementation using the library

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor 
```

```python
# Example dataset
data = {
    "area_sqft": [1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800],
    "bedrooms":  [2, 2, 3, 3, 4, 4, 5, 5],
    "house_size": [980, 1180, 1480, 1790, 1980, 2180, 2480, 2780],
    "age":       [20, 15, 10, 12, 8, 5, 3, 2]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
#Calculate the VIF
vif = pd.DataFrame()
vif["feature"] = df.columns
vif["VIF"] = [
    variance_inflation_factor(df.values, i)
    for i in range(df.shape[1])
]

print(vif)
```

# So, area_sqft and house_size has extremely high multicollinearity. We simply dropped the columns to remove the multi-collinearity


```python
#So, removing the one columns
df = df.drop('house_size', axis = 1)
vif = pd.DataFrame()
vif['feature'] = df.columns
vif["VIF"] = [
    variance_inflation_factor(df.values, i)
    for i in range(df.shape[1])
]

print(vif)
```

```python
print(df[['area_sqft','bedrooms','age']].corr())
```

```python
df_reduced = df.drop("bedrooms", axis=1)
```

```python
X = df_reduced[["area_sqft", "age"]]

vif = pd.DataFrame()
vif["feature"] = X.columns
vif["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

print(vif)
```

# Final Results

Initially, the model showed severe multicollinearity, with very high VIF values for area_sqft, bedrooms, and house_size. After removing house_size, area_sqft and bedrooms still had very high VIF values, indicating strong multicollinearity between these variables. bedrooms was subsequently removed. The final VIF values for area_sqft and age were approximately 1.78, indicating that multicollinearity was no longer a significant issue.
