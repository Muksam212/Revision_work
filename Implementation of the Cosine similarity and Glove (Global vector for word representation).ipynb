# Implementation of the cosine similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
a = np.array([[1, 2, 3]])
b = np.array([[2, 4, 6]])
```

```python
similarity = cosine_similarity(a, b)
```

# Use GLOVE also to capture the relationship between texts

It stands for Global vectors for word representations.

```python
import gensim.downloader as api

glove = api.load("glove-wiki-gigaword-50")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

uncle = glove["uncle"]
aunt = glove["aunt"]

similarity = cosine_similarity([uncle], [aunt])

print(similarity[0][0])
```

This means the Glove vector for uncle and aunt is fairly similar.
