# Gradient Descent

Gradient descent is an iterative optimisation algorithm used to find the minimum of a function, most commonly a cost or loss function in machine learning.

The formula for gradient descent:

# Gradient Descent

## Mean Squared Error

The linear regression model is:

$$
\hat{y}_i = wx_i + b
$$

The Mean Squared Error is:

$$
J(w,b) = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

### Gradient with respect to $w$

$$
\frac{\partial J}{\partial w}
=
-\frac{2}{n}
\sum_{i=1}^{n}
x_i(y_i-\hat{y}_i)
$$

### Gradient with respect to $b$

$$
\frac{\partial J}{\partial b}
=
-\frac{2}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)
$$

### Gradient Descent Updates

$$
w_{\text{new}}
=
w_{\text{old}}
-
\alpha
\frac{\partial J}{\partial w}
$$

$$
b_{\text{new}}
=
b_{\text{old}}
-
\alpha
\frac{\partial J}{\partial b}
$$


Where,

- α is the learning rate, a hyperparameter (a tuning setting chosen before training) that determines the size of the steps a machine learning model takes when updating its parameters during optimisation.

```python
#Gradient Descent
x = 10 #Starting point
learning_rate = 0.01 
iterations = 100 #how many times you want to loop

for i in range(iterations):
    # Gradient of f(x) = x^2
    gradient = 2 * x
    #update the x
    x = x - learning_rate * gradient
    print(f"Iteration {i+1}: x = {x:.4f}")
```

You'll see x gradually move toward 0, which is the minimum of 
x
2
.

# A more useful using the: Linear Regression

Suppose:

The formula:
$$
y = wx + b
$$

We want to find the best w and b

```python
import numpy as np
# Data
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

#Initial Parameters
w = 0
b = 0

# Hyperparameter (a tuning setting chosen before training) that determines 
# the size of the steps a machine learning model takes when updating its parameters during optimisation.
learning_rate = 0.01 
epochs = 100 #How many times you want to iterate

n = len(X)

for epoch in range(epochs):

    # Prediction
    y_pred = w * X + b

    # Mean Squared Error gradients
    dw = (-2/n) * np.sum(X * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    # Gradient descent update
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("w =", w)
print("b =", b)
```

So the model learns:

$$
y
≈
2
x
$$
