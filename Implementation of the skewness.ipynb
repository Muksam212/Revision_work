# Skewness 

- Positive skewness (right-skewed distribution): Lower values with higher density, and higher values with lower density. (Mode < Median < Mean)
- Negative skewness (left-skewed distribution): Lower value with lower density, and higher value with higher density. (Mean < Median < Mode).
- Zero skewness: The data is perfectly balanced, meaning the left and right sides of the distribution curve are exact mirror images of each other.

# Implementation

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from scipy.stats import skew, mode
```

```python
# Positive (right) skewed data
positive = np.random.exponential(scale=2, size=1000)

# Negative (left) skewed data
negative = -np.random.exponential(scale=2, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, data, title in zip(
    axes,
    [positive, negative],
    ["Positive (Right) Skewness", "Negative (Left) Skewness"]
):
    mean = np.mean(data)
    median = np.median(data)
    mode_value = mode(data, keepdims=True).mode[0]
    skewness = skew(data)

    ax.hist(data, bins=30, density=True, alpha=0.7, edgecolor="black")

    ax.axvline(mean, color="red", label=f"Mean = {mean:.2f}")
    ax.axvline(median, color="green", label=f"Median = {median:.2f}")
    ax.axvline(mode_value, color="blue", label=f"Mode = {mode_value:.2f}")

    ax.set_title(f"{title}\nSkewness = {skewness:.2f}")
    ax.set_xlabel("Value")
    ax.set_ylabel("Density")
    ax.legend()

plt.tight_layout()
plt.show()
```

Final Results shows

- Positive Skewness: Lower values with higher density and higher values with lower density (Mode < Median < Mean)
- Negative Skewness: Lower values with lower density and higher values with higher density (Mean < Median < Mode)

# Implementation of zero-skewed distribution

```python
# Symmetric data
data = np.random.normal(loc=50, scale=10, size=1000)

mean = np.mean(data)
median = np.median(data)
skewness = skew(data)

print("Mean:", mean)
print("Median:", median)
print("Skewness:", skewness)

plt.hist(data, bins=30, density=True, edgecolor="black", alpha=0.7)

plt.axvline(mean, color="red", label=f"Mean = {mean:.2f}")
plt.axvline(median, color="green", linestyle="--",
            label=f"Median = {median:.2f}")

plt.title(f"Zero/Approximately Zero Skewness = {skewness:.2f}")
plt.legend()
plt.show()
```

In zero skewness: The data is perfectly balanced, meaning the left and right sides of the distribution curve are exact mirror images of each other.
