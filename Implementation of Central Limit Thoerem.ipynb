# CLT (Central Limit Theorem)

- The Central Limit Theorem says that when we take many random samples from a population, the distribution of the sample means becomes approximately normal as the sample size gets larger, even if the original population is not normally distributed.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(42) #To make the result reproducible

# Generate 10,000 samples
# Each sample contains 30 dice rolls

sample_means = [] #empty list to append to later
for i in range(10000):
    sample = np.random.randint(1, 7, size = 30)
    sample_means.append(np.mean(sample))

# Plot distribution of sample means
#
plt.hist(sample_means, bins=30, edgecolor='black')
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Central Limit Theorem")
plt.show()
```

You should see a distribution that is approximately bell-shaped, even though the original die-roll distribution is not normal.

Why is this useful?

The CLT is important because it allows us to use methods based on the normal distribution when working with sample means, which is the foundation for things like:

- Confidence Intervals
- Hypothesis testing
- t-tests
- ANOVA
