# Chi-squared Test and Cramer's V
$$
\chi^2 = \sum \frac{(O-E)^2}{E}
$$


Where,

- O => Observed,
- E => Expected



For Cramer's V

$$
V = \sqrt{\frac{\chi^2}{n(k-1)}}
$$

To find the relationship between two categorical variables, but it will not show how strong the variables. For that,
we use Cramer's v, it tell is there any association between them.


What the Boundaries Mean
• 0.0: Indicates no association at all; the variables are completely independent.
• 1.0: Indicates a perfect association; knowing the category of one variable allows you to predict the other with 100% certainty
